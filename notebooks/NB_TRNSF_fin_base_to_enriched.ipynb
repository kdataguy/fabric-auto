# Base to Enriched

Apply finance-domain enrichment rules and write the enriched Delta table.

```python
source_table = "base.source_data"
target_table = "enriched.source_data"
amount_column = "amount"
```

```python
from pyspark.sql import functions as F

base_df = spark.table(source_table)
enriched_df = base_df.withColumn("_enriched_at", F.current_timestamp())
if amount_column in enriched_df.columns:
    enriched_df = enriched_df.withColumn(amount_column, F.col(amount_column).cast("decimal(18,2)"))
    enriched_df = enriched_df.withColumn("amount_is_valid", F.col(amount_column).isNotNull())

enriched_df.write.mode("overwrite").format("delta").saveAsTable(target_table)
print(f"Wrote {enriched_df.count()} rows to {target_table}")
```
