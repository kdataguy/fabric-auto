# Initialize Bronze Metadata

Create the control tables used by metadata-driven ingestion. Run once per environment.

```python
from pyspark.sql import types as T

config_schema = T.StructType([
    T.StructField("source_name", T.StringType(), False),
    T.StructField("source_format", T.StringType(), False),
    T.StructField("source_path", T.StringType(), False),
    T.StructField("target_table", T.StringType(), False),
    T.StructField("load_type", T.StringType(), False),
    T.StructField("watermark_column", T.StringType(), True),
    T.StructField("enabled", T.BooleanType(), False),
]) 
run_log_schema = T.StructType([
    T.StructField("source_name", T.StringType(), False),
    T.StructField("target_table", T.StringType(), False),
    T.StructField("run_started_at", T.StringType(), False),
    T.StructField("status", T.StringType(), False),
    T.StructField("row_count", T.LongType(), False),
])
```

```python
config_rows = [("example_source", "json", "Files/incoming/finance", "raw.source_data", "incremental", "updated_at", True)]
config_df = spark.createDataFrame(config_rows, config_schema)
config_df.write.mode("overwrite").format("delta").saveAsTable("raw._sys_ingestion_config")

empty_run_log = spark.createDataFrame([], run_log_schema)
empty_run_log.write.mode("append").format("delta").saveAsTable("raw._sys_run_log")
print("Metadata tables initialized. Add or edit source rows before running ingestion.")
```
